## 5. Pre-processing – Missing Value Handling

Missing values can affect data analysis and machine learning models.

We first identify the missing values using `isnull().sum()`.

```python
# Check missing values in each column

print("Missing values in each column:")
print(df.isnull().sum())
```

## 6. Handling Missing Values

Missing values are handled using suitable replacement methods.

- Numerical missing values in `age` are filled using the median.
- Categorical missing values in `embarked` are filled using the mode.
- Missing values in `deck` are replaced with `"Unknown"`.

```python
# Handle missing values

df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

# Add "Unknown" as a category before filling missing deck values
df["deck"] = df["deck"].cat.add_categories("Unknown")
df["deck"] = df["deck"].fillna("Unknown")

print("Missing values handled successfully!")
print(df.isnull().sum())
```

## 7. Label Encoding

Label encoding converts categorical values into numerical values so that they can be used by machine learning algorithms.

For example:
- Male → 1
- Female → 0

```python
import pandas as pd
import seaborn as sns

# Load Titanic dataset
df = sns.load_dataset("titanic")

# Handle missing values
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

# Convert deck to object and handle missing values
df["deck"] = df["deck"].astype("object")
df["deck"] = df["deck"].fillna("Unknown")

print("Dataset loaded and missing values handled successfully!")
```

```python
from sklearn.preprocessing import LabelEncoder

# Create a copy of the dataset for encoding
encoded_df = df.copy()

# Apply label encoding to sex
label_encoder = LabelEncoder()
encoded_df["sex"] = label_encoder.fit_transform(encoded_df["sex"])

print("Label encoding completed!")
print(encoded_df[["sex"]].head())
```

## 8. Encoded Data Verification
