## Task 5: Data Manipulation with Pandas

### Objective
To understand Pandas and perform different data manipulation and analysis operations using Series and DataFrames.

```python
import pandas as pd
import numpy as np

print("Pandas imported successfully!")
print("Pandas Version:", pd.__version__)
```

## 2. Key Data Structures in Pandas

### Series
A Series is a one-dimensional labeled array.

### DataFrame
A DataFrame is a two-dimensional labeled table.

```python
# Creating a Pandas Series

marks = pd.Series([85, 90, 78, 92])

print("Pandas Series:")
print(marks)
```

```python
# Creating a Pandas DataFrame

data = {
    "Name": ["Aarati", "Priya", "Neha", "Sneha"],
    "Age": [21, 22, 20, 23],
    "Marks": [85, 90, 78, 92]
}

df = pd.DataFrame(data)

print("Pandas DataFrame:")
print(df)
```

## 3. Creating Data Manually

Pandas allows us to create data manually using `pd.Series()` and `pd.DataFrame()`.

```python
# Creating a Series manually

numbers = pd.Series([10, 20, 30, 40, 50])

print("Manual Series:")
print(numbers)
```

```python
# Creating a DataFrame manually

students = pd.DataFrame({
    "Name": ["Aarati", "Priya", "Neha", "Sneha"],
    "Age": [21, 22, 20, 23],
    "Marks": [85, 90, 78, 92]
})

print("Manual DataFrame:")
print(students)
```

## 4. Data Operations – Indexing and Slicing

Indexing is used to select specific columns or rows from a DataFrame.

- `df['col']` selects a column.
- `df.iloc[]` selects rows and columns using numerical positions.
- `df.loc[]` selects rows and columns using labels.

```python
# Selecting the Marks column

print("Marks Column:")
print(df["Marks"])
```

```python
# Selecting rows using iloc

print("First two rows:")
print(df.iloc[0:2])
```

```python
# Selecting first three rows

print(df[0:3])
```

## 5. Data Cleaning

Data cleaning is used to find and handle missing values in a dataset.

- `.isnull()` checks for missing values.
- `.dropna()` removes rows containing missing values.
- `.fillna()` fills missing values.

```python
# Creating a DataFrame with missing values

messy_data = pd.DataFrame({
    "Name": ["Aarati", "Priya", "Neha", "Sneha"],
    "Age": [21, np.nan, 20, 23],
    "Marks": [85, 90, np.nan, 92]
})

print("Data with missing values:")
print(messy_data)
```

```python
# Check missing values

print("Missing values:")
print(messy_data.isnull())
```

```python
# Count missing values in each column

print("Missing value count:")
print(messy_data.isnull().sum())
```

```python
# Remove rows with missing values

print("After dropna():")
print(messy_data.dropna())
```

```python
# Fill missing values

cleaned_data = messy_data.fillna({
    "Age": messy_data["Age"].mean(),
    "Marks": messy_data["Marks"].mean()
})

print("After fillna():")
print(cleaned_data)
```

## 6. Renaming Columns

The `rename()` function is used to change column names in a DataFrame.

```python
# Rename columns

renamed_df = df.rename(columns={
    "Name": "Student_Name",
    "Marks": "Score"
})

print("Renamed DataFrame:")
print(renamed_df)
```

## 7. Filtering Data

Filtering is used to select rows that satisfy a condition.

```python
# Filter students with marks greater than 80

print("Students with marks greater than 80:")
print(df[df["Marks"] > 80])
```

## 8. Aggregation and Grouping

`groupby()` is used to group data based on a column.

Common aggregation functions are:
- `.sum()` – calculates total
- `.mean()` – calculates average
- `.count()` – counts values

```python
# Creating a sales DataFrame

sales = pd.DataFrame({
    "Category": ["Electronics", "Electronics", "Clothing", "Clothing", "Food"],
    "Sales": [1000, 1500, 800, 700, 500],
    "Quantity": [5, 7, 4, 3, 10]
})

grouped = sales.groupby("Category")

print("Total Sales:")
print(grouped["Sales"].sum())

print("\nAverage Sales:")
print(grouped["Sales"].mean())

print("\nCount:")
print(grouped["Sales"].count())
```

## 9. Merging and Joining

Pandas can combine data from different DataFrames using `pd.merge()` and `pd.concat()`.

```python
# Creating two DataFrames

students_info = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Aarati", "Priya", "Neha"]
})

student_marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 90, 78]
})

# Merge the DataFrames using ID

merged_df = pd.merge(students_info, student_marks, on="ID")

print("Merged DataFrame:")
print(merged_df)
```

```python
# Creating two DataFrames

part1 = pd.DataFrame({
    "Name": ["A", "B"],
    "Score": [80, 85]
})

part2 = pd.DataFrame({
    "Name": ["C", "D"],
    "Score": [90, 95]
})

# Concatenate the DataFrames

concatenated_df = pd.concat([part1, part2], ignore_index=True)

print("Concatenated DataFrame:")
print(concatenated_df)
```

## 10. File Input and Output

Pandas can read and write CSV and Excel files.

- `pd.read_csv()` – read a CSV file
- `df.to_csv()` – save data to a CSV file
- `pd.read_excel()` – read an Excel file
- `df.to_excel()` – save data to an Excel file

```python
# Creating sample data

io_df = pd.DataFrame({
    "Name": ["Aarati", "Priya", "Neha"],
    "Score": [85, 90, 78]
})

# Save DataFrame to CSV

io_df.to_csv("task5_sample.csv", index=False)

# Read CSV file

csv_data = pd.read_csv("task5_sample.csv")

print("CSV data:")
print(csv_data)
```

```python
# Save DataFrame to Excel

io_df.to_excel("task5_sample.xlsx", index=False)

# Read Excel file

excel_data = pd.read_excel("task5_sample.xlsx")

print("Excel data:")
print(excel_data)
```

## 11. Time Series

Time series data contains values recorded over dates or time.

- `pd.to_datetime()` converts values into datetime format.
- `.resample()` groups time-based data into a specified frequency.

```python
# Creating time series data

dates = pd.date_range("2026-01-01", periods=6, freq="D")

time_df = pd.DataFrame({
    "Date": dates,
    "Sales": [100, 120, 150, 130, 160, 180]
})

# Convert Date column to datetime

time_df["Date"] = pd.to_datetime(time_df["Date"])

# Set Date as index

time_df = time_df.set_index("Date")

print("Time Series Data:")
print(time_df)
```

```python
# Calculate weekly sales

weekly_sales = time_df.resample("W")["Sales"].sum()

print("Weekly Sales:")
print(weekly_sales)
```

## 13. Condition Selection – AND and OR

In Pandas:
- `&` is used for AND.
- `|` is used for OR.

Each condition should be written inside parentheses.

```python
# AND condition

print("Students with marks greater than 80 AND age less than 23:")
print(df[(df["Marks"] > 80) & (df["Age"] < 23)])
```

```python
# OR condition

print("Students with marks greater than 90 OR age less than 21:")
print(df[(df["Marks"] > 90) | (df["Age"] < 21)])
```
