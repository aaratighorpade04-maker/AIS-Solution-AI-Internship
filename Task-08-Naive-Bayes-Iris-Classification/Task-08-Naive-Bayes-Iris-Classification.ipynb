```python
import pandas as pd
import numpy as np

from sklearn.datasets import load_iris

print("Libraries imported successfully!")
```

## 1. Load the Iris Dataset

The Iris dataset is a commonly used dataset for classification.

It contains 150 observations and 4 numerical features:

- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

The target variable contains three classes:

- Setosa
- Versicolor
- Virginica

```python
# Load Iris dataset

iris = load_iris()

# Create DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add target column
df["target"] = iris.target

print("First 5 rows:")
print(df.head())

print("\nDataset Shape:", df.shape)
```

## 2. Separate Independent and Dependent Variables

The independent variables (X) are the four features of the Iris dataset.

The dependent variable (y) is the target class representing the Iris species.

### Independent Variables (X)
- Sepal Length
- Sepal Width
- Petal Length
- Petal Width

### Dependent Variable (y)
- Target / Iris Species

```python
# Separate independent and dependent variables

X = df.drop("target", axis=1)
y = df["target"]

print("Independent Variables (X):")
print(X.head())

print("\nDependent Variable (y):")
print(y.head())
```

```python
print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
```

## 3. Train-Test Split

The dataset is divided into training and testing sets using an 80:20 ratio.

- 80% of the data is used for training.
- 20% of the data is used for testing.

The training data is used to learn patterns from the dataset, while the testing data is used to evaluate how well the trained model performs on unseen data.

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

## 4. Naive Bayes Classification

Naive Bayes is a supervised machine learning classification algorithm based on **Bayes' Theorem**.

It is called "Naive" because it assumes that all input features are conditionally independent of each other given the class.

### Bayes' Theorem

P(C|X) = [P(X|C) × P(C)] / P(X)

Where:

- P(C|X) = Probability of class C given features X
- P(X|C) = Probability of features X given class C
- P(C) = Prior probability of class C
- P(X) = Probability of the features

The classifier predicts the class having the highest posterior probability.

### General Naive Bayes Formula

For multiple features:

P(C|X₁, X₂, ..., Xₙ) ∝ P(C) × P(X₁|C) × P(X₂|C) × ... × P(Xₙ|C)

### Main Naive Bayes Types

1. **Gaussian Naive Bayes** – Used for continuous numerical data.
2. **Bernoulli Naive Bayes** – Used for binary data, such as 0/1 or True/False.
3. **Multinomial Naive Bayes** – Commonly used for count-based data, such as word counts in text classification.

### Hand Calculation Example

Suppose we have two classes: A and B.

P(A) = 0.6  
P(X|A) = 0.5  

P(B) = 0.4  
P(X|B) = 0.2  

Calculate the scores:

Score(A) = P(A) × P(X|A)

= 0.6 × 0.5

= 0.30

Score(B) = P(B) × P(X|B)

= 0.4 × 0.2

= 0.08

Since 0.30 > 0.08, the observation is classified as Class A.

This is a simple example of how Naive Bayes compares probabilities to select a class.

## 5. Gaussian Naive Bayes

Gaussian Naive Bayes assumes that the numerical features follow a Gaussian (normal) distribution within each class.

### Gaussian Probability Formula

P(x|C) = 1 / √(2πσ²) × e^[-(x-μ)²/(2σ²)]

Where:

- x = Feature value
- μ = Mean of the feature for the class
- σ² = Variance of the feature for the class
- C = Class

Gaussian Naive Bayes calculates the probability of each class and predicts the class with the highest probability.

```python
from sklearn.naive_bayes import GaussianNB

# Create Gaussian Naive Bayes model
gnb = GaussianNB()

print("Gaussian Naive Bayes model created successfully!")
```

```python
# Train Gaussian Naive Bayes model

gnb.fit(X_train, y_train)

print("Gaussian Naive Bayes model trained successfully!")
```

```python
# Predict the test data

y_pred_gnb = gnb.predict(X_test)

print("Predicted values:")
print(y_pred_gnb)
```

## 6. Accuracy of Gaussian Naive Bayes

Accuracy measures how many predictions made by the model are correct.

### Formula

Accuracy = Number of Correct Predictions / Total Number of Predictions

Accuracy Score = Correct Predictions / Total Predictions × 100

```python
from sklearn.metrics import accuracy_score

# Calculate accuracy
accuracy_gnb = accuracy_score(y_test, y_pred_gnb)

print("Gaussian Naive Bayes Accuracy:", accuracy_gnb)
print("Gaussian Naive Bayes Accuracy (%):", accuracy_gnb * 100)
```

## 7. Bernoulli Naive Bayes

Bernoulli Naive Bayes is used when features are binary in nature.

Examples:

- 0 / 1
- True / False
- Present / Absent

The Iris dataset contains continuous numerical features, so the features are converted into binary values before applying BernoulliNB.

In this implementation:

- Values greater than the feature median → 1
- Values less than or equal to the feature median → 0

```python
from sklearn.naive_bayes import BernoulliNB

# Convert continuous features into binary features
thresholds = X_train.median()

X_train_bernoulli = (X_train > thresholds).astype(int)
X_test_bernoulli = (X_test > thresholds).astype(int)

print("Binary Training Data:")
print(X_train_bernoulli.head())
```

```python
# Create Bernoulli Naive Bayes model
bnb = BernoulliNB()

# Train the model
bnb.fit(X_train_bernoulli, y_train)

print("Bernoulli Naive Bayes model trained successfully!")
```

```python
# Predict test data

y_pred_bnb = bnb.predict(X_test_bernoulli)

print("BernoulliNB Predictions:")
print(y_pred_bnb)
```

## 8. Accuracy of Bernoulli Naive Bayes

The accuracy score is calculated by comparing the predicted values with the actual test values.

### Formula

Accuracy = Correct Predictions / Total Predictions × 100

```python
# Calculate BernoulliNB accuracy

accuracy_bnb = accuracy_score(y_test, y_pred_bnb)

print("Bernoulli Naive Bayes Accuracy:", accuracy_bnb)
print("Bernoulli Naive Bayes Accuracy (%):", accuracy_bnb * 100)
```

```python
correct_bnb = np.sum(y_test.values == y_pred_bnb)
total_bnb = len(y_test)

print("Correct Predictions:", correct_bnb)
print("Total Predictions:", total_bnb)
print("Accuracy:", (correct_bnb / total_bnb) * 100, "%")
```

## 9. Multinomial Naive Bayes

Multinomial Naive Bayes is commonly used for discrete or count-based features.

It is widely used in applications such as:

- Text classification
- Document classification
- Spam detection
- Word-count based classification

The Iris dataset contains continuous values, so we cannot directly treat the original measurements as multinomial count data.

For this implementation, the Iris features are converted into non-negative discrete values using binning.

```python
from sklearn.naive_bayes import MultinomialNB

# Convert continuous Iris features into non-negative discrete values
X_train_multi = (X_train * 10).round().astype(int)
X_test_multi = (X_test * 10).round().astype(int)

print("Multinomial Training Data:")
print(X_train_multi.head())
```

```python
# Create Multinomial Naive Bayes model
mnb = MultinomialNB()

# Train the model
mnb.fit(X_train_multi, y_train)

print("Multinomial Naive Bayes model trained successfully!")
```

```python
# Predict the test data

y_pred_mnb = mnb.predict(X_test_multi)

print("MultinomialNB Predictions:")
print(y_pred_mnb)
```

## 10. Accuracy of Multinomial Naive Bayes

Accuracy is calculated by comparing the predicted classes with the actual classes in the test dataset.

### Formula

Accuracy = Correct Predictions / Total Predictions × 100

```python
# Calculate MultinomialNB accuracy

accuracy_mnb = accuracy_score(y_test, y_pred_mnb)

print("Multinomial Naive Bayes Accuracy:", accuracy_mnb)
print("Multinomial Naive Bayes Accuracy (%):", accuracy_mnb * 100)
```

```python
correct_mnb = np.sum(y_test.values == y_pred_mnb)
total_mnb = len(y_test)

print("Correct Predictions:", correct_mnb)
print("Total Predictions:", total_mnb)
print("Accuracy:", (correct_mnb / total_mnb) * 100, "%")
```

## 11. Comparison of Naive Bayes Models

The three Naive Bayes algorithms implemented in this task are:

- Gaussian Naive Bayes
- Bernoulli Naive Bayes
- Multinomial Naive Bayes

Their accuracy scores are compared to understand their performance on the Iris dataset.

## 13. Naive Bayes Hand Calculation

Naive Bayes uses Bayes' theorem to calculate the probability of a class.

### Bayes' Theorem

P(C|X) = [P(X|C) × P(C)] / P(X)

Where:

- P(C|X) = Posterior probability
- P(X|C) = Likelihood
- P(C) = Prior probability
- P(X) = Evidence

For comparing different classes, P(X) is common, so we can compare:

P(C|X) ∝ P(X|C) × P(C)

### Hand Calculation Example

Suppose there are two classes: Class A and Class B.

For a new observation X:

P(A) = 0.6
P(X|A) = 0.5

P(B) = 0.4
P(X|B) = 0.2

#### For Class A

Score(A) = P(X|A) × P(A)

Score(A) = 0.5 × 0.6

Score(A) = 0.30

#### For Class B

Score(B) = P(X|B) × P(B)

Score(B) = 0.2 × 0.4

Score(B) = 0.08

Since:

0.30 > 0.08

The predicted class is **Class A**.

### Gaussian Naive Bayes Formula

For continuous numerical features, GaussianNB uses:

P(x|C) = 1 / √(2πσ²) × e^[-(x-μ)²/(2σ²)]

Where:

- x = observed feature value
- μ = mean
- σ² = variance
- C = class

The probabilities of the features are combined with the class prior probability to determine the predicted class.
