## 1. Numbers in Python

Python mainly provides these numeric types:

| Type | Example | Description |
|---|---:|---|
| `int` | `10` | Whole numbers without a decimal point |
| `float` | `10.5` | Numbers containing a decimal point |
| `complex` | `3 + 4j` | Numbers containing a real and imaginary part |

Python's built-in `type()` function can be used to check the type of a value.


````python
# Integer: a whole number
age = 21

# Float: a number with a decimal point
height = 5.6

# Complex number: real part + imaginary part
z = 3 + 4j

print("age =", age, "| type =", type(age))
print("height =", height, "| type =", type(height))
print("z =", z, "| type =", type(z))
````

### Basic arithmetic operations

Python supports:
- Addition: `+`
- Subtraction: `-`
- Multiplication: `*`
- Division: `/`
- Floor division: `//`
- Modulus/remainder: `%`
- Exponentiation: `**`


````python
# Two numbers for arithmetic operations
a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** 2)
````

### Arithmetic with complex numbers

````python
# Complex number arithmetic
c1 = 2 + 3j
c2 = 1 + 2j

print("Complex addition:", c1 + c2)
print("Complex subtraction:", c1 - c2)
print("Complex multiplication:", c1 * c2)

# The conjugate changes the sign of the imaginary part
print("Conjugate of c1:", c1.conjugate())
````

## 2. Operators in Python

Operators are symbols or keywords used to perform operations on values.

### Main operator categories
1. **Arithmetic:** `+`, `-`, `*`, `/`, `//`, `%`, `**`
2. **Comparison:** `==`, `!=`, `<`, `>`, `<=`, `>=`
3. **Logical:** `and`, `or`, `not`
4. **Assignment:** `=`, `+=`, `-=`, `*=`, `/=`, etc.


### 2.1 Arithmetic operators

````python
x = 15
y = 4

print("x + y =", x + y)   # Addition
print("x - y =", x - y)   # Subtraction
print("x * y =", x * y)   # Multiplication
print("x / y =", x / y)   # Division
print("x // y =", x // y) # Floor division
print("x % y =", x % y)   # Modulus
print("x ** y =", x ** y) # Exponentiation
````

### 2.2 Comparison operators

Comparison operators return either `True` or `False`.


````python
x = 10
y = 20

print("x == y:", x == y)  # Equal to
print("x != y:", x != y)  # Not equal to
print("x < y:", x < y)    # Less than
print("x > y:", x > y)    # Greater than
print("x <= y:", x <= y)  # Less than or equal to
print("x >= y:", x >= y)  # Greater than or equal to
````

### 2.3 Logical operators

- `and` → `True` only when both conditions are true.
- `or` → `True` when at least one condition is true.
- `not` → reverses a Boolean result.


````python
age = 22
has_id = True

# Both conditions must be True
print("Can enter (and):", age >= 18 and has_id)

# At least one condition must be True
print("Eligible by either condition (or):", age >= 18 or has_id)

# not reverses True to False and False to True
print("Not has_id:", not has_id)
````

### 2.4 Assignment operators

````python
# Start with a value
number = 10
print("Initial value:", number)

number += 5   # Same as: number = number + 5
print("After += 5:", number)

number -= 3   # Same as: number = number - 3
print("After -= 3:", number)

number *= 2   # Same as: number = number * 2
print("After *= 2:", number)

number /= 4   # Same as: number = number / 4
print("After /= 4:", number)
````

## 3. Lists in Python

A **list** is an ordered, mutable collection. Lists can contain multiple values and can be changed after creation.

Example:
```python
fruits = ["apple", "banana", "mango"]
```


````python
# Creating a list
fruits = ["apple", "banana", "mango", "orange", "grapes"]

print("List:", fruits)
print("Type:", type(fruits))
print("Length:", len(fruits))
````

### 3.1 Accessing elements — indexing

Python uses **zero-based indexing**, so the first element has index `0`.


````python
fruits = ["apple", "banana", "mango", "orange", "grapes"]

print("First element:", fruits[0])
print("Second element:", fruits[1])
print("Last element:", fruits[-1])
print("Second-last element:", fruits[-2])
````

### 3.2 Accessing elements — slicing

Syntax:
```python
list[start:stop:step]
```

The `stop` position is excluded.


````python
fruits = ["apple", "banana", "mango", "orange", "grapes"]

print("First three:", fruits[0:3])
print("From index 2:", fruits[2:])
print("Up to index 3:", fruits[:3])
print("Every second element:", fruits[::2])
print("Reversed using slicing:", fruits[::-1])
````

### 3.3 Modifying an element

Because lists are mutable, an element can be changed using its index.


````python
fruits = ["apple", "banana", "mango"]

# Change the value at index 1
fruits[1] = "watermelon"

print("Updated list:", fruits)
````

### 3.4 Adding elements — `append()` and `insert()`

- `append()` adds an element to the end.
- `insert(index, value)` adds an element at a specific position.


````python
fruits = ["apple", "banana", "mango"]

# Add one item at the end
fruits.append("orange")
print("After append():", fruits)

# Add an item at index 1
fruits.insert(1, "grapes")
print("After insert():", fruits)
````

### 3.5 Removing elements — `remove()`, `pop()`, and `clear()`

- `remove(value)` removes the first matching value.
- `pop(index)` removes and returns an element. If no index is given, it removes the last element.
- `clear()` removes all elements.


````python
# remove()
fruits = ["apple", "banana", "mango", "banana"]
fruits.remove("banana")
print("After remove('banana'):", fruits)

# pop()
removed_item = fruits.pop(1)
print("Removed item:", removed_item)
print("After pop(1):", fruits)

# clear()
fruits.clear()
print("After clear():", fruits)
````

### 3.6 Other useful list methods

- `sort()` → sorts the list.
- `reverse()` → reverses the list.
- `extend()` → adds all elements from another iterable.
- `count()` → counts occurrences of a value.
- `index()` → returns the position of the first matching value.
- `copy()` → creates a shallow copy of the list.


````python
numbers = [5, 2, 8, 2, 9, 1]

# sort() arranges values in ascending order
numbers.sort()
print("After sort():", numbers)

# reverse() reverses the current order
numbers.reverse()
print("After reverse():", numbers)

# extend() adds multiple values
numbers.extend([10, 11])
print("After extend():", numbers)

# count() tells how many times a value occurs
print("Count of 2:", numbers.count(2))

# index() returns the first index of a value
print("Index of 2:", numbers.index(2))

# copy() creates a separate list
numbers_copy = numbers.copy()
print("Copied list:", numbers_copy)
````

### Mini practice: list operations

````python
# Practice example using student marks
marks = [78, 85, 92, 67, 85]

print("Original marks:", marks)

# Add a new mark
marks.append(88)

# Change the first mark
marks[0] = 80

# Count how many times 85 occurs
print("Number of students with 85:", marks.count(85))

# Sort the marks
marks.sort()

print("Final marks:", marks)
````

## 4. Tuples in Python

A **tuple** is an ordered collection that is **immutable**, meaning its elements cannot be changed after creation.

Example:
```python
colors = ("red", "green", "blue")
```

Tuples support indexing, slicing, concatenation, repetition, `count()`, and `index()`.


````python
# Creating a tuple
colors = ("red", "green", "blue", "yellow", "blue")

print("Tuple:", colors)
print("Type:", type(colors))
print("Length:", len(colors))
````

### 4.1 Accessing tuple elements — indexing and slicing

````python
colors = ("red", "green", "blue", "yellow", "blue")

# Indexing
print("First element:", colors[0])
print("Last element:", colors[-1])

# Slicing
print("First three:", colors[:3])
print("Elements from index 2:", colors[2:])
print("Every second element:", colors[::2])
````

### 4.2 Concatenation using `+`

Two tuples can be combined using the `+` operator.


````python
tuple1 = ("Python", "Java")
tuple2 = ("C", "C++")

combined = tuple1 + tuple2

print("Tuple 1:", tuple1)
print("Tuple 2:", tuple2)
print("Combined tuple:", combined)
````

### 4.3 Repetition using `*`

````python
languages = ("Python", "Java")

# Repeat the tuple three times
repeated = languages * 3

print("Original:", languages)
print("Repeated:", repeated)
````

### 4.4 `count()` and `index()`

````python
numbers_tuple = (10, 20, 10, 30, 10, 40)

# count() counts how many times 10 occurs
print("Count of 10:", numbers_tuple.count(10))

# index() returns the first position of 30
print("Index of 30:", numbers_tuple.index(30))
````

## 5. Sets in Python

A **set** is an unordered collection of unique elements.

Important properties:
- Duplicate values are automatically removed.
- Sets do not use normal positional indexing.
- Sets are useful for mathematical set operations and removing duplicates.


````python
# Creating a set
numbers = {1, 2, 3, 3, 4, 5}

print("Set:", numbers)
print("Type:", type(numbers))
````

### 5.1 Adding and removing elements

````python
numbers = {1, 2, 3}

# add() adds one element
numbers.add(4)
print("After add(4):", numbers)

# remove() removes a specified element
numbers.remove(2)
print("After remove(2):", numbers)
````

### 5.2 Union, intersection, and difference

Suppose:
- Set A contains students who know Python.
- Set B contains students who know Java.

Then:
- **Union** = students in A or B.
- **Intersection** = students common to A and B.
- **Difference** = elements in one set but not the other.


````python
python_students = {"Aarav", "Priya", "Rahul", "Sneha"}
java_students = {"Rahul", "Sneha", "Vikram", "Neha"}

# Union: all unique elements from both sets
print("Union:", python_students.union(java_students))

# Intersection: elements common to both sets
print("Intersection:", python_students.intersection(java_students))

# Difference: Python students who are not in Java set
print("Python only:", python_students.difference(java_students))

# Difference in the opposite direction
print("Java only:", java_students.difference(python_students))
````

## 6. Dictionaries in Python

A **dictionary** stores data as **key-value pairs**.

Example:
```python
student = {
    "name": "Aarati",
    "age": 21,
    "course": "E&TC"
}
```

Keys should be unique. Values can be accessed using their keys.


````python
# Creating a dictionary
student = {
    "name": "Aarati",
    "age": 21,
    "course": "Electronics and Telecommunication",
    "cgpa": 8.87
}

print("Dictionary:", student)
print("Type:", type(student))
````

### 6.1 Accessing values

You can access a value using `dictionary[key]` or safely using `get(key)`.


````python
student = {
    "name": "Aarati",
    "age": 21,
    "course": "E&TC"
}

print("Name:", student["name"])

# get() safely returns the value
print("Age:", student.get("age"))

# If the key does not exist, get() returns None by default
print("City:", student.get("city"))

# You can provide a custom default value
print("City:", student.get("city", "Not available"))
````

### 6.2 `keys()`, `values()`, and `items()`

````python
student = {
    "name": "Aarati",
    "age": 21,
    "course": "E&TC"
}

# keys() returns all keys
print("Keys:", student.keys())

# values() returns all values
print("Values:", student.values())

# items() returns key-value pairs
print("Items:", student.items())

# Loop through all key-value pairs
for key, value in student.items():
    print(key, ":", value)
````

### 6.3 Adding and updating dictionary items

Use `dictionary[key] = value` to add or change one item.

Use `update()` to add or update multiple items.


````python
student = {
    "name": "Aarati",
    "age": 21
}

# Add a new key-value pair
student["city"] = "Kolhapur"
print("After adding city:", student)

# Change an existing value
student["age"] = 22
print("After updating age:", student)

# Add/update multiple values
student.update({
    "course": "E&TC",
    "skill": "Python"
})
print("After update():", student)
````

### 6.4 Removing dictionary items using `pop()`

````python
student = {
    "name": "Aarati",
    "age": 21,
    "city": "Kolhapur",
    "skill": "Python"
}

# pop() removes the specified key and returns its value
removed_value = student.pop("city")

print("Removed value:", removed_value)
print("Dictionary after pop():", student)
````

## 7. Common Python Errors

Errors are normal while learning programming. Understanding the error message helps you find and fix the problem.

The common errors covered here are:
- `IndentationError`
- `NameError`
- `ValueError`
- `TypeError`
- `IndexError`
- `KeyError`


### 7.1 IndentationError

**Cause:** Python uses indentation to define blocks of code. Incorrect indentation can produce an `IndentationError`.

**Incorrect idea:**
```python
if True:
print("Hello")
```

The `print()` statement must be indented.

**Correct version:**


````python
# Correct indentation
if True:
    print("Hello, Python!")
````

### 7.2 NameError

**Cause:** A variable or function name is used before it has been defined, or the spelling is incorrect.

**Example of the problem:**
```python
print(student_name)
```

If `student_name` was never created, Python raises `NameError`.

**Fix:** Define the variable before using it.


````python
# Correct example
student_name = "Aarati"

print("Student name:", student_name)
````

### 7.3 ValueError

**Cause:** The type of value may be acceptable, but the actual value is not valid for the requested operation.

For example, converting `"hello"` to an integer is invalid.

**Fix:** Validate the input before conversion or handle the exception.


````python
# Safe handling of ValueError
user_input = "25"

try:
    number = int(user_input)
    print("Converted number:", number)
except ValueError:
    print("Please enter a valid integer.")
````

### 7.4 TypeError

**Cause:** An operation is performed on an inappropriate data type.

For example, adding a string and an integer directly is not allowed.

**Fix:** Convert values to compatible types or use the correct operation.


````python
# Correct handling by converting the integer to a string
name = "Aarati"
age = 21

message = name + " is " + str(age) + " years old."
print(message)
````

### 7.5 IndexError

**Cause:** An index outside the valid range of a list, tuple, or similar sequence is accessed.

For a list of length 3, valid indexes are `0`, `1`, and `2`.

**Fix:** Check the length or use a valid index.


````python
# Safe index handling
numbers = [10, 20, 30]

index = 2

if index < len(numbers):
    print("Value:", numbers[index])
else:
    print("Index is outside the list.")
````

### 7.6 KeyError

**Cause:** A dictionary is accessed with a key that does not exist.

**Fix:** Use `get()` when a key may be missing, or check with `in`.


````python
student = {
    "name": "Aarati",
    "age": 21
}

# Safe way to access a possibly missing key
print("City:", student.get("city", "Key not found"))

# Another safe method
if "city" in student:
    print("City:", student["city"])
else:
    print("The 'city' key does not exist.")
````

## 8. Final Combined Practice

The following example combines lists, tuples, sets, dictionaries, numbers, and operators in one small student-data example.


````python
# Student marks stored in a list
marks = [78, 85, 92, 85, 74]

# Calculate basic statistics using built-in functions
total_marks = sum(marks)
number_of_subjects = len(marks)
average_marks = total_marks / number_of_subjects

print("Marks:", marks)
print("Total:", total_marks)
print("Number of subjects:", number_of_subjects)
print("Average:", average_marks)

# Use a tuple for fixed student information
student_info = ("Aarati", "E&TC", 2026)
print("Student information:", student_info)

# Use a set to find unique marks
unique_marks = set(marks)
print("Unique marks:", unique_marks)

# Use a dictionary for structured student data
student = {
    "name": student_info[0],
    "branch": student_info[1],
    "passing_year": student_info[2],
    "average_marks": average_marks
}

print("Student dictionary:", student)

# Comparison operator
if average_marks >= 75:
    print("Result: Good performance")
else:
    print("Result: Needs improvement")
````
