## Task 4: Data Analysis with NumPy

This notebook provides a practical introduction to NumPy and demonstrates arrays, array operations, reshaping, indexing, mathematical functions, random numbers, and broadcasting.

## 1. Introduction to NumPy
NumPy (Numerical Python) is a Python library used for fast numerical computations. It is widely used in data science, machine learning, and scientific computing.

Installation command:
`pip install numpy`

```python
import numpy as np
print("NumPy version:", np.__version__)
```

## 2. Arrays in NumPy
NumPy uses an N-dimensional array called `ndarray`.

```python
arr = np.array([1, 2, 3, 4, 5])
zeros = np.zeros(5)
ones = np.ones(5)
arange = np.arange(0, 10, 2)
linspace = np.linspace(0, 1, 5)

print("Array:", arr)
print("Zeros:", zeros)
print("Ones:", ones)
print("Arange:", arange)
print("Linspace:", linspace)
print("Index 2:", arr[2])
print("Slice 1:4:", arr[1:4])
```

## 3. Array Operations
NumPy supports arithmetic operations, element-wise functions, and broadcasting.

```python
a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Square root:", np.sqrt(a))
print("Sine:", np.sin(a))
print("Exponential:", np.exp([1, 2, 3]))
```

## 4. Array Shapes and Reshaping
`.shape` gives the size of each dimension and `.ndim` gives the number of dimensions.

```python
arr2 = np.arange(1, 13)
matrix = arr2.reshape(3, 4)

print("Matrix:\n", matrix)
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Flattened:", matrix.flatten())
print("Transpose:\n", matrix.transpose())
```

```python
x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])

print("Concatenate:\n", np.concatenate((x, y)))
print("Vertical stack:\n", np.vstack((x, y)))
print("Horizontal stack:\n", np.hstack((x, y)))
print("Split:\n", np.split(np.array([1, 2, 3, 4]), 2))
```

## 5. Array Manipulation
Indexing, slicing, masking, and fancy indexing help select specific elements from arrays.

```python
data = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print("Element [1,2]:", data[1, 2])
print("First two rows:\n", data[:2, :])
print("Values greater than 50:", data[data > 50])
indices = np.array([0, 2])
print("Fancy indexing:", data[indices])
```

## 6. Mathematical Functions
NumPy provides universal functions (ufuncs), statistical functions, and linear algebra functions.

```python
p = np.array([10, 20, 30])
q = np.array([1, 2, 3])

print("np.add:", np.add(p, q))
print("np.subtract:", np.subtract(p, q))
print("np.multiply:", np.multiply(p, q))

stats = np.array([10, 20, 30, 40, 50])
print("Mean:", np.mean(stats))
print("Median:", np.median(stats))
print("Standard deviation:", np.std(stats))
print("Variance:", np.var(stats))
print("Minimum:", np.min(stats))
print("Maximum:", np.max(stats))
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Dot product:\n", np.dot(A, B))
print("Matrix multiplication:\n", np.matmul(A, B))
print("Inverse of A:\n", np.linalg.inv(A))
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

## 7. Random Number Generation
NumPy can generate random numbers, integers, samples, shuffles, and permutations. A seed makes random results reproducible.

```python
np.random.seed(42)

print("Random numbers:", np.random.rand(5))
print("Random integers:", np.random.randint(1, 20, size=5))

sample = np.random.choice(np.arange(1, 11), size=5, replace=False)
print("Random sample:", sample)

numbers = np.arange(1, 6)
np.random.shuffle(numbers)
print("Shuffled array:", numbers)
print("Random permutation:", np.random.permutation(np.arange(1, 6)))
```

## 8. Broadcasting
Broadcasting allows NumPy to perform operations between arrays with compatible different shapes by automatically expanding dimensions when possible.

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
row = np.array([10, 20, 30])

print("Matrix:\n", matrix)
print("Row array:", row)
print("After broadcasting:\n", matrix + row)
```

### Broadcasting Rules
1. Compare dimensions from right to left.
2. Two dimensions are compatible when they are equal or one of them is 1.
3. If dimensions are compatible, NumPy broadcasts the smaller array across the larger array.

In the example above, the row `[10, 20, 30]` is added to every row of the 2×3 matrix.
