## Task 3: Python Functions

This notebook demonstrates Python functions for maximum value, list sum, distinct elements, combinations, permutations, vowel/consonant counting, case conversion, and case counting.

## 1. Function to Find Maximum


```python
def find_maximum(numbers):
    return max(numbers)

numbers = [10, 25, 7, 42, 18]
print("Maximum value is =", find_maximum(numbers))
```

## 2. Function to Sum All Numbers in a List


```python
def sum_list(numbers):
    total = 0
    for number in numbers:
        total += number
    return total

sample_list = [8, 2, 3, 0, 7]
print("Sum is =", sum_list(sample_list))
```

## 3. Function to Return Distinct Elements


```python
def distinct_elements(items):
    unique = []
    for item in items:
        if item not in unique:
            unique.append(item)
    return unique

sample_list = [1, 2, 3, 3, 3, 3, 4, 5]
print("Unique List :", distinct_elements(sample_list))
```

## 4. Function to Find Total Number of Combinations

For `a, b, c` choosing 2 elements, the combinations are `ab, ac, bc`, so the total is 3.

```python
from itertools import combinations

def total_combinations(items, r=2):
    result = list(combinations(items, r))
    return len(result), result

items = ['a', 'b', 'c']
count, result = total_combinations(items, 2)
print("Total combinations =", count)
print("Combinations =", [''.join(x) for x in result])
```

## 5. Function to Find Total Number of Permutations

For `a, b, c` choosing 2 elements, the permutations are `ab, ac, ba, bc, ca, cb`, so the total is 6.

```python
from itertools import permutations

def total_permutations(items, r=2):
    result = list(permutations(items, r))
    return len(result), result

items = ['a', 'b', 'c']
count, result = total_permutations(items, 2)
print("Total permutations =", count)
print("Permutations =", [''.join(x) for x in result])
```

## 6. Function to Count Vowels and Consonants


```python
def count_vowels_consonants(word):
    vowels = 0
    consonants = 0
    for char in word.lower():
        if char.isalpha():
            if char in 'aeiou':
                vowels += 1
            else:
                consonants += 1
    return vowels, consonants

word = "statistics"
vowels, consonants = count_vowels_consonants(word)
print("Count of vowel is =", vowels)
print("Count of consonant is =", consonants)
```

## 7. Function to Convert Lowercase Words to Uppercase


```python
def to_uppercase(word):
    return word.upper()

word = "statistics"
print("Result is =", to_uppercase(word))
```

## 8. Function to Count Lowercase and Uppercase Letters


```python
def count_case(text):
    uppercase = 0
    lowercase = 0
    for char in text:
        if char.isupper():
            uppercase += 1
        elif char.islower():
            lowercase += 1
    return uppercase, lowercase

text = "STatiStiCS"
upper, lower = count_case(text)
print("Upper case :", upper)
print("Lower case :", lower)
```
