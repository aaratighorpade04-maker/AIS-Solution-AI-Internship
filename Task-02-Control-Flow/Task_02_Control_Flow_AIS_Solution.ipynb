# Task 2.a — If-Else

Conditional statements allow Python to make decisions.

### Main forms
- `if` — executes code when a condition is true.
- `if-else` — chooses between two alternatives.
- `if-elif-else` — chooses between multiple conditions.

### Important correction for the grade problem
The intended grade ranges are interpreted as:
- `O`: marks > 75
- `A`: 60 < marks <= 75
- `B`: 50 < marks <= 60
- `C`: 35 <= marks <= 50
- `Fail`: marks < 35

The boundaries are written explicitly to avoid overlapping conditions.


## 1) Grade Code

````python
# Grade classification using if-elif-else
p = 50

if p > 75:
    grade = "O"
elif 60 < p <= 75:
    grade = "A"
elif 50 < p <= 60:
    grade = "B"
elif 35 <= p <= 50:
    grade = "C"
else:
    grade = "Fail"

print("Marks:", p)
print("Grade:", grade)
````

### Grade code with different marks

````python
# Test the grade logic with multiple marks
marks_list = [82, 70, 55, 40, 25]

for p in marks_list:
    if p > 75:
        grade = "O"
    elif 60 < p <= 75:
        grade = "A"
    elif 50 < p <= 60:
        grade = "B"
    elif 35 <= p <= 50:
        grade = "C"
    else:
        grade = "Fail"

    print(p, "->", grade)
````

## 2) Check Whether a Number Is Divisible by 2 or 3

Possible outputs:
1. Divisible by both 2 and 3
2. Divisible by 2, but not 3
3. Divisible by 3, but not 2
4. Not divisible by either 2 or 3


````python
n = 7

if n % 2 == 0 and n % 3 == 0:
    print("Divisible by 2 and 3")
elif n % 2 == 0:
    print("Divisible by 2, 3 not")
elif n % 3 == 0:
    print("Divisible by 3, 2 not")
else:
    print("Not divisible by 2 and 3")
````

# Task 2.b — For Loop

A `for` loop repeats a block of code for each item in a sequence or for each value produced by `range()`.

### Useful `range()` forms
- `range(stop)` → starts at 0 and stops before `stop`.
- `range(start, stop)` → starts at `start`, stops before `stop`.
- `range(start, stop, step)` → moves by `step`.


## 1) Print Odd Values Between 20 and 80 Without Using If

````python
# range(21, 81, 2) starts at the first odd number after 20
# and increases by 2, so no if statement is required.
for number in range(21, 81, 2):
    print(number, end=" ")
````

## 2) Create a List of 1 to 20 Using a For Loop

````python
numbers = []

# range(1, 21) generates 1 through 20
for number in range(1, 21):
    numbers.append(number)

print(numbers)
````

## 3) Create a List of 20 to 1 Without Using Reverse

````python
numbers = []

# A negative step creates the numbers in descending order.
for number in range(20, 0, -1):
    numbers.append(number)

print(numbers)
````

## 4) Cube of Odd Values Between 20 and 40

````python
for number in range(21, 41, 2):
    cube = number ** 3
    print("Number:", number, "Cube:", cube)
````

## 5) Print Friends' Names and Corresponding Ages

````python
name = ["a", "b", "c", "d", "e"]
age = [20, 21, 23, 25, 24]

# zip() pairs the first name with the first age, and so on.
for friend_name, friend_age in zip(name, age):
    print("my name is", friend_name + ",", "my age is", friend_age)
````

# Task 2.c — Solve Problems Using If, For Loops, and Data Type Methods

This section combines conditions, loops, and Python's built-in methods/functions.


## 1) Swap the First and Last Element of a List

````python
numbers = [10, 20, 30, 40, 50]

# Save the first element temporarily.
# Then assign the last element to the first position.
# Finally assign the saved first element to the last position.
numbers[0], numbers[-1] = numbers[-1], numbers[0]

print("After swapping:", numbers)
````

## 2) Count the Length of a String

````python
text = "Python Programming"

# len() returns the number of characters, including spaces.
length = len(text)

print("String:", text)
print("Length:", length)
````

## 3) Sum Only Non-Negative Integers

````python
numbers = [1, 4, -5, -20, 10]

total = 0

for number in numbers:
    # Add the number only when it is zero or positive.
    if number >= 0:
        total += number

print("Sum of non-negative integers:", total)
````

## 4) Factorial

````python
number = 3
factorial = 1

# 3! = 3 * 2 * 1
for i in range(number, 0, -1):
    factorial *= i

print("Factorial of", number, "=", factorial)
````

### Factorial with input validation

````python
number = 5

if number < 0:
    print("Factorial is not defined for negative numbers.")
else:
    factorial = 1

    for i in range(1, number + 1):
        factorial *= i

    print("Factorial of", number, "=", factorial)
````

# Task 2.d — While Loop

A `while` loop repeatedly executes a block of code as long as its condition is `True`.

### General structure
```python
while condition:
    # code
    # update the condition variable
```

Always make sure the loop variable changes when necessary; otherwise, the loop may never stop.


## 1) Odd-Even Using While Loop

````python
number = 1

while number <= 10:
    if number % 2 == 0:
        print(number, "is Even")
    else:
        print(number, "is Odd")

    number += 1
````

## 2) Print the Star Pattern Using While Loop

````python
rows = 5
i = 1

while i <= rows:
    # Multiply the string by i to create the pattern.
    print("* " * i)
    i += 1
````

## 3) Create List 1 to 20 Using While Loop

````python
numbers = []
number = 1

while number <= 20:
    numbers.append(number)
    number += 1

print(numbers)
````

## 4) Create List 20 to 1 Using While Loop

````python
numbers = []
number = 20

while number >= 1:
    numbers.append(number)
    number -= 1

print(numbers)
````

## 5) Control Statements — break, continue, pass

````python
# break: immediately stops the loop
for number in range(1, 11):
    if number == 6:
        break
    print(number)

print("Loop stopped at 6.")
````

````python
# continue: skips the current iteration and moves to the next one
for number in range(1, 11):
    if number % 2 == 0:
        continue
    print("Odd:", number)
````

````python
# pass: does nothing; it is useful as a placeholder
for number in range(1, 6):
    if number == 3:
        pass
    print(number)
````

## Additional While Loop Examples

### Example 1 — Countdown

````python
number = 5

while number >= 1:
    print(number)
    number -= 1

print("Go!")
````

### Example 2 — Sum of 1 to 10

````python
number = 1
total = 0

while number <= 10:
    total += number
    number += 1

print("Sum:", total)
````

### Example 3 — Print Multiplication Table

````python
number = 5
i = 1

while i <= 10:
    print(number, "x", i, "=", number * i)
    i += 1
````

### Example 4 — Find the First Multiple of 7 Greater Than 20

````python
number = 21

while number % 7 != 0:
    number += 1

print("First multiple of 7 greater than 20:", number)
````

### Example 5 — Reverse Digits of a Number

````python
number = 12345
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number //= 10

print("Reversed number:", reverse)
````

# Final Practice — Combined Control Flow

This example uses a list, a `for` loop, an `if-elif-else` condition, and a dictionary to classify numbers.


````python
numbers = [5, 12, 18, 25, 31, 42, 50]

result = {
    "even": [],
    "odd": []
}

for number in numbers:
    if number % 2 == 0:
        result["even"].append(number)
    else:
        result["odd"].append(number)

print("Even numbers:", result["even"])
print("Odd numbers:", result["odd"])
````
